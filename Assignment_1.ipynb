## 1. Plot the four input data samples. Label the axes and title as appropriate.

```python
from matplotlib import pyplot as plt

dataSample = [(0, 0), (0, 1), (1, 0), (1, 1)]
x = [sample[0] for sample in dataSample]
y = [sample[1] for sample in dataSample]

plt.plot(x, y, '-')
plt.title('Input Data Sample')
plt.xlabel('Input Sample (x)')
plt.ylabel('Input Sample (y)')
plt.show()
```

 ## 2. Write a function that takes two arguments as an input and performs AND operation. The function returns the result to the main program.


```python
def andArgument(x, y):
  if x == 1 & y == 1:
    return True
  else:
    return False
```

## 3. Write two additional functions like in part (b); each would perform OR and XOR operations and return the result to the main program.


```python
def orArgument(x, y):
  if x == 0 & y == 0:
    return False
  else:
    return True

```

```python
def xorArgument(x, y):
  if x != y:
    return True
  else:
    return False
```

## 4. Create a for loop to call the three functions each for data sample each time.


```python
for sample in dataSample:
  print(andArgument(sample[0], sample[1]))
  print(orArgument(sample[0], sample[1]))
  print(xorArgument(sample[0], sample[1]))
```

## 5. Store the output of each function in a separate list/array.

```python
andArgumentLIST = []
orArgumentLIST = []
xorArgumentLSIT = []

for sample in dataSample:
  andArgumentLIST.append(andArgument(sample[0], sample[1]))
  orArgumentLIST.append(orArgument(sample[0], sample[1]))
  xorArgumentLSIT.append(xorArgument(sample[0], sample[1]))
```

## 6. Print out the results for the three functions at the end of your program.


```python
print(andArgumentLIST)
print(orArgumentLIST)
print(xorArgumentLSIT)
```
